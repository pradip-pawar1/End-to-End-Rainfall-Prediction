# Rainfall prediction Final ML model

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/processed.csv')
```

```python
df.head()
```

```python
# Split data into X & y

X = df.drop('RainTomorrow', axis = 1)
y = df['RainTomorrow']
```

```python
# Save model as pkl

import pickle
from xgboost import XGBClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", XGBClassifier(
        subsample=0.8,
        n_estimators=200,
        max_depth=8,
        learning_rate=0.05,
        colsample_bytree=0.8,
        random_state=42
    ))
])

pipeline.fit(X, y)


model_package = {
    "model": pipeline,
    "threshold": 0.45,
    "feature_names": list(X.columns)
}
# Save as .pkl file

with open("rain_model_3.pkl", "wb") as f:
    pickle.dump(model_package, f)
```

```python
# Confirming Model Inside Pickle Is Pipeline
import pickle
with open("rain_model_3.pkl", "rb") as f:
    package = pickle.load(f)

model = package["model"]

print(type(model))
```
